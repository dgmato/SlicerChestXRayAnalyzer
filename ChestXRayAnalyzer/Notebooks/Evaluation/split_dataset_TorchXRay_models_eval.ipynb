```python
import os
import pandas as pd
import numpy as np
import shutil
from sklearn.model_selection import train_test_split
import random
```

## Prepare the dataset

```python
csv_path = 'C:/Users/maria/OneDrive/Documentos/Escritorio/CodeChestXRay/TrainXray/Data/PADCHEST_chest_x_ray_images_labels_160K_01.02.19.csv'
```

```python
# Read the CSV file
csv_table = pd.read_csv(csv_path, low_memory = False)

# Replace missing values in "Labels" column with an empty string
csv_table['Labels'].fillna('', inplace = True)

#---------------------Filters--------------------: 
imageIDs_available = "C:/Users/maria/OneDrive/Documentos/Escritorio/CodeChestXRay/TrainXray/normal_classification/normal_classify_training/Resources/txt_files/PC512_Full_PA_filtered_dataset/imageIDs/PadChest_512_PA_manually_filtered_87946.txt"

with open(imageIDs_available, "r") as f:
    avail_imgID = [line.strip() for line in f.readlines()] 
    
# Filter by available image ID
csv_table = csv_table[csv_table["ImageID"].isin(avail_imgID)]
print("Images PA in PadChest512 (excluding the labels “exclude” o “suboptimal study”"+
      f"and manually filtered): {len(csv_table)}")
```

## Filter each pathology

```python
contain_infiltrations =            ["infiltrates",
                                   "interstitial pattern",
                                   "ground glass pattern",
                                   "reticular interstitial pattern",
                                   "reticulonodular interstitial pattern",
                                   "alveolar pattern",
                                   "consolidation",
                                   "air bronchogram"]
```

```python
# Filter the rows where the Labels column contains a label
atelectasis = csv_table[csv_table['Labels'].str.contains('atelectasis')]
infiltration = csv_table[csv_table['Labels'].str.contains('|'.join(contain_infiltrations))]
pneumothorax = csv_table[csv_table['Labels'].str.contains('pneumothorax')]
effusion = csv_table[csv_table['Labels'].str.contains('effusion')]
pneumonia = csv_table[csv_table['Labels'].str.contains('pneumonia')]
cardiomegaly = csv_table[csv_table['Labels'].str.contains('cardiomegaly')]
nodule = csv_table[csv_table['Labels'].str.contains('nodule')]
mass = csv_table[csv_table['Labels'].str.contains('mass')]
fracture = csv_table[csv_table['Labels'].str.contains('fracture')]
```

### Select images that do not have the pathology

```python
# Filter the rows where the Labels column contains a label
not_atelectasis = csv_table[~csv_table['Labels'].str.contains('atelectasis')]
not_infiltration = csv_table[~csv_table['Labels'].str.contains('|'.join(contain_infiltrations))]
not_pneumothorax = csv_table[~csv_table['Labels'].str.contains('pneumothorax')]
not_effusion = csv_table[~csv_table['Labels'].str.contains('effusion')]
not_pneumonia = csv_table[~csv_table['Labels'].str.contains('pneumonia')]
not_cardiomegaly = csv_table[~csv_table['Labels'].str.contains('cardiomegaly')]
not_nodule = csv_table[~csv_table['Labels'].str.contains('nodule')]
not_mass = csv_table[~csv_table['Labels'].str.contains('mass')]
not_fracture = csv_table[~csv_table['Labels'].str.contains('fracture')]
```

```python
print(f"not_atelectasis: {len(not_atelectasis)}\n"+ 
      f"not_infiltration: {len(not_infiltration)}\n" +
      f"not_pneumothorax: {len(not_pneumothorax)}\n" +
      f"not_effusion: {len(not_effusion)}\n" + 
      f"not_pneumonia: {len(not_pneumonia)}\n" +
      f"not_cardiomegaly: {len(not_cardiomegaly)}\n" +
      f"not_nodule: {len(not_nodule)}\n" +
      f"not_mass: {len(not_mass)}\n" +
      f"not_fracture: {len(not_fracture)}\n")
```

### Sample the images to have the same number of positive/negative cases in each pathology

```python
not_atelectasis = not_atelectasis.sample(len(atelectasis)) 
not_infiltration = not_infiltration.sample(len(infiltration))
not_pneumothorax = not_pneumothorax.sample(len(pneumothorax))
not_effusion = not_effusion.sample(len(effusion))
not_pneumonia = not_pneumonia.sample( len(pneumonia))
not_cardiomegaly = not_cardiomegaly.sample(len(cardiomegaly))
not_nodule = not_nodule.sample(len(nodule))
not_mass = not_mass.sample(len(mass))
not_fracture = not_fracture.sample(len(fracture))
```

## Create a txt file with the labels for each pathology

```python
dataframes = {
    'atelectasis': (atelectasis, not_atelectasis),
    'infiltration': (infiltration, not_infiltration),
    'pneumothorax': (pneumothorax, not_pneumothorax),
    'effusion': (effusion, not_effusion),
    'pneumonia': (pneumonia, not_pneumonia),
    'cardiomegaly': (cardiomegaly, not_cardiomegaly),
    'nodule': (nodule, not_nodule),
    'mass': (mass, not_mass),
    'fracture': (fracture, not_fracture),
}

for label, (df, not_df) in dataframes.items():
    image_ids = df['ImageID'].to_numpy()
    negated_image_ids = not_df['ImageID'].to_numpy()
    all_image_ids = np.concatenate((image_ids, negated_image_ids))
    np.savetxt(f'{label}_IDs.txt', all_image_ids, fmt='%s')
```
